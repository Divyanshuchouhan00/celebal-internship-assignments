```python
Q1 = df['Estimated_Deliveries'].quantile(0.25)
Q3 = df['Estimated_Deliveries'].quantile(0.75)

IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = df[
    (df['Estimated_Deliveries'] < lower_bound)  |
    (df['Estimated_Deliveries'] > upper_bound)
]

print("Number of Outliers:", len(outliers))

plt.scatter(
    df.index,
    df['Estimated_Deliveries'],
    label='Normal Data'
)

plt.scatter(
    outliers.index,
    outliers['Estimated_Deliveries'],
    label='Outliers'
)

plt.title("Outlier Detection in Estimated Deliveries")

plt.xlabel("Data Points")
plt.ylabel("Estimated Deliveries")

plt.legend()

plt.show()
```

Observation:
A total of 12 outliers were detected using the IQR method. These values  may represent fluctuations in Tesla deliveries.

```python
#Feature Engineering

encoder = LabelEncoder()

df['Region'] = encoder.fit_transform(df['Region'])
df['Model'] = encoder.fit_transform(df['Model'])
df['Source_Type'] = encoder.fit_transform(df['Source_Type'])

print("Categorical Columns Encoded")
```

```python
#Features and Target Variable

x = df.drop('Estimated_Deliveries', axis=1)

y = df['Estimated_Deliveries']
```

```python
#Train_Test_Split the data

X_train, X_test, y_train, y_test = train_test_split(
    x,
    y,
    test_size=0.2,
    random_state=42
)

print("Training Shape:", X_train.shape)
print("Testing Shape:", X_test.shape)
```

```python
#Linear Regression Model

lr_model = LinearRegression()

lr_model.fit(X_train, y_train)

lr_pred = lr_model.predict(X_test)

print("linear Regression Model Trained")
```

```python
#Ridge Regression Model

ridge_model = Ridge(alpha=1.0)

ridge_model.fit(X_train, y_train)

ridge_pred = ridge_model.predict(X_test)
```

```python
#Lasso Regression Model

lasso_model = Lasso(alpha=1.0)

lasso_model.fit(X_train, y_train)

lasso_pred = lasso_model.predict(X_test)
```

```python
comparison_df = pd.DataFrame({

    "Model": ["Linear Regression", "Ridge", "Lasso"],

    "MAE": [
        mean_absolute_error(y_test, lr_pred),
        mean_absolute_error(y_test, ridge_pred),
        mean_absolute_error(y_test, lasso_pred)
    ],

    "RMSE": [
        np.sqrt(mean_squared_error(y_test, lr_pred)),
        np.sqrt(mean_squared_error(y_test, ridge_pred)),
        np.sqrt(mean_squared_error(y_test, lasso_pred))
    ],

    "R2 Score": [
        r2_score(y_test, lr_pred),
        r2_score(y_test, ridge_pred),
        r2_score(y_test, lasso_pred)
    ]

})

comparison_df
```

```python
#Model Comparison of 3 models

plt.figure(figsize=(8,5))

plt.bar(
    comparison_df["Model"],
    comparison_df["R2 Score"]
)

plt.ylabel("R2 Score")
plt.title("Model Comparison (R2 Score)")


best_model = comparison_df.loc[
    comparison_df["R2 Score"].idxmax()
]

print("Best Model:", best_model["Model"])
print("Best R2 Score:", round(best_model["R2 Score"],6))

plt.show()
```

```python
#HyperParameter  Tuning

#RIDGE:
ridge_params = {
    'alpha' : [0.01, 0.1, 1, 10, 100]
}

ridge_grid = GridSearchCV(
    Ridge(),
    ridge_params,
    cv=3
)

ridge_grid.fit(X_train, y_train)

print("Best Ridge Alpha:")
print(ridge_grid.best_params_)

#Lasso:
lasso_params = {
    'alpha': [0.01, 0.1, 1, 10, 100]
}

lasso_grid = GridSearchCV(
    Lasso(),
    lasso_params,
    cv=3
)

lasso_grid.fit(X_train, y_train)

print("Best Lasso Alpha:")
print(lasso_grid.best_params_)
```

```python
#Time series Forecasting

yearly_data = df.groupby("Year")['Estimated_Deliveries'].sum()

model = ARIMA(yearly_data, order=(1,1,1))

model_fit = model.fit()

forecast = model_fit.forecast(steps=5)

future_years = np.arange(
    yearly_data.index.max() +1,
    yearly_data.index.max() +6
)

forecast_df = pd.DataFrame({
    'Year': future_years,
    'Forecasted_Deliveries': forecast.values.astype(int)
})

forecast_df
```
